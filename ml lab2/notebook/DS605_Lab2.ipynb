TASK 2  Vectorized Arithmetic and Linear Algebra 

1. Create two matrices

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

Matrix addition

```python
addition = A + B

print("A + B:")
print(addition)
```

Element-wise multiplication

```python
element_multiplication = A * B

print("A * B:")
print(element_multiplication)
```

Matrix multiplication

```python
matrix_multiplication = A @ B

print("A @ B:")
print(matrix_multiplication)
```

2. Transpose and determinant

```python
transpose = A.T

print("Transpose of A:")
print(transpose)

determinant = np.linalg.det(A)

print("\nDeterminant of A:", determinant)
```

3. Inverse and verification

```python
if not np.isclose(np.linalg.det(A), 0):
    A_inverse = np.linalg.inv(A)

    print("Inverse of A:")
    print(A_inverse)

    verification = A @ A_inverse

    print("\nA @ A^-1:")
    print(verification)

    print("\nIs approximately identity?",
          np.allclose(verification, np.eye(A.shape[0])))
```

3. Compare different imputation methods

Mean

```python
df_mean = df.copy()

print("Missing Age before:",
      df_mean['Age'].isnull().sum())

mean_age = df_mean['Age'].mean()

df_mean['Age'] = df_mean['Age'].fillna(mean_age)

print("Missing Age after:",
      df_mean['Age'].isnull().sum())
```

mean

```python
df_mean = df.copy()

df_mean['Age'] = df_mean['Age'].fillna(
    df_mean['Age'].mean()
)

print("Mean imputation missing:",
      df_mean['Age'].isnull().sum())
```

Median

```python
df_median = df.copy()

df_median['Age'] = df_median['Age'].fillna(
    df_median['Age'].median()
)

print("Median imputation missing:",
      df_median['Age'].isnull().sum())
```

Mode

```python
df_mode = df.copy()

mode_age = df_mode['Age'].mode()[0]

df_mode['Age'] = df_mode['Age'].fillna(mode_age)

print("Mode used:", mode_age)
print("Mode imputation missing:",
      df_mode['Age'].isnull().sum())
```

Random value imputation

```python
import numpy as np
import pandas as pd

# Make a copy of the original dataframe
df_random = df.copy()

# Set random seed
np.random.seed(42)

# Find missing Age values
missing_indices = df_random["Age"].isnull()

# Select random existing Age values
random_values = np.random.choice(
    df_random["Age"].dropna(),
    size=missing_indices.sum(),
    replace=True
)

# Fill missing Age values
df_random.loc[missing_indices, "Age"] = random_values

# Check missing values after imputation
print("Random imputation missing:",
      df_random["Age"].isnull().sum())
```

4. Fare Outliers using IQR

```python
Q1 = df['Fare'].quantile(0.25)
```

```python
Q3 = df['Fare'].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

Find outliers

```python
fare_outliers = df[
    (df['Fare'] < lower_bound) |
    (df['Fare'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("\nNumber of Fare outliers:",
      len(fare_outliers))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation = df[numerical_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

Find strongest positive and negative correlations

```python
corr_pairs = correlation.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) !=
    corr_pairs.index.get_level_values(1)
]

print("Strongest positive correlation:")
print(corr_pairs.idxmax(), corr_pairs.max())

print("\nStrongest negative correlation:")
print(corr_pairs.idxmin(), corr_pairs.min())
```
