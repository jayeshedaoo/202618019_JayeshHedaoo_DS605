Creating the preprocessing pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numerical_features),
    ("cat", categorical_transformer, categorical_features)
])
```

```python
X_reg_train = X_reg.loc[train_idx]
X_reg_test = X_reg.loc[test_idx]

y_reg_train = y_reg.loc[train_idx]
y_reg_test = y_reg.loc[test_idx]

X_cls_train = X_cls.loc[train_idx]
X_cls_test = X_cls.loc[test_idx]

y_cls_train = y_cls.loc[train_idx]
y_cls_test = y_cls.loc[test_idx]
```

```python
print("Categorical features:", categorical_features)
print("Numerical features:", numerical_features)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import time
import numpy as np
```

```python
regression_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])
```

```python
start_time = time.perf_counter()

regression_model.fit(X_reg_train, y_reg_train)

regression_train_time = time.perf_counter() - start_time

print("Regression training time:", regression_train_time, "seconds")
```

```python
start_time = time.perf_counter()

y_reg_pred = regression_model.predict(X_reg_test)

regression_prediction_time = time.perf_counter() - start_time

print("Regression prediction time:", regression_prediction_time, "seconds")
```

```python
sklearn_regression_results = {
    "Model": "Scikit-learn Linear Regression",
    "MAE": mae,
    "RMSE": rmse,
    "R2": r2,
    "Training Time": regression_train_time,
    "Prediction Time": regression_prediction_time
}

print(sklearn_regression_results)
```

```python
# Part B setup

import numpy as np
import pandas as pd
import time

# Load the dataset again
df = pd.read_csv("data/garments_worker_productivity.csv")

# Create classification target
df["MeetsTarget"] = (
    df["actual_productivity"] >= df["targeted_productivity"]
).astype(int)

# Define features
regression_features = [
    col for col in df.columns
    if col not in ["actual_productivity", "MeetsTarget"]
]

classification_features = regression_features.copy()

# Identify categorical and numerical features
categorical_features = df[regression_features].select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = df[regression_features].select_dtypes(
    exclude=["object"]
).columns.tolist()

# Recreate the SAME 80/20 split using the same random seed.
# This is only for reproducing the fixed split;
# no sklearn train-test utility is used here.
rng = np.random.RandomState(42)
indices = np.arange(len(df))
rng.shuffle(indices)

test_size = int(np.ceil(len(df) * 0.2))

test_idx = indices[:test_size]
train_idx = indices[test_size:]

# Create manual copy
manual_df = df.copy()

manual_features = regression_features.copy()

manual_X = manual_df[manual_features].copy()

manual_y_reg = manual_df["actual_productivity"].copy()
manual_y_cls = manual_df["MeetsTarget"].copy()

print("Dataset shape:", df.shape)
print("Training samples:", len(train_idx))
print("Testing samples:", len(test_idx))
print("Categorical features:", categorical_features)
print("Numerical features:", numerical_features)
```

```python
manual_X = pd.get_dummies(
    manual_X,
    columns=categorical_features,
    drop_first=False
)

manual_X = manual_X.astype(float)

print("Shape after encoding:", manual_X.shape)
print(manual_X.head())
```

```python
manual_X_train = manual_X.iloc[train_idx].copy()
manual_X_test = manual_X.iloc[test_idx].copy()

y_train_reg_manual = manual_y_reg.iloc[train_idx].to_numpy()
y_test_reg_manual = manual_y_reg.iloc[test_idx].to_numpy()

y_train_cls_manual = manual_y_cls.iloc[train_idx].to_numpy()
y_test_cls_manual = manual_y_cls.iloc[test_idx].to_numpy()

print("Training shape:", manual_X_train.shape)
print("Testing shape:", manual_X_test.shape)
```

```python
for column in manual_X_train.columns:
    median_value = manual_X_train[column].median()

    manual_X_train[column] = manual_X_train[column].fillna(median_value)
    manual_X_test[column] = manual_X_test[column].fillna(median_value)
```

```python
print("Missing values in training data:",
      manual_X_train.isnull().sum().sum())

print("Missing values in testing data:",
      manual_X_test.isnull().sum().sum())
```

```python
train_mean = manual_X_train.mean()
train_std = manual_X_train.std()

train_std = train_std.replace(0, 1)
```

```python
manual_X_train_scaled = (
    manual_X_train - train_mean
) / train_std

manual_X_test_scaled = (
    manual_X_test - train_mean
) / train_std
```

```python
X_train_manual = manual_X_train_scaled.to_numpy()
X_test_manual = manual_X_test_scaled.to_numpy()
```

```python
print("Final training shape:", X_train_manual.shape)
print("Final testing shape:", X_test_manual.shape)
```

Linear Regression from scratch

```python
X_train_lr = np.c_[
    np.ones(X_train_manual.shape[0]),
    X_train_manual
]

X_test_lr = np.c_[
    np.ones(X_test_manual.shape[0]),
    X_test_manual
]

print("Training matrix:", X_train_lr.shape)
print("Testing matrix:", X_test_lr.shape)
```

```python
start_time = time.perf_counter()

linear_weights = np.linalg.lstsq(
    X_train_lr,
    y_train_reg_manual,
    rcond=None
)[0]

manual_regression_train_time = time.perf_counter() - start_time

print("Manual Linear Regression training time:",
      manual_regression_train_time,
      "seconds")
```

```python
start_time = time.perf_counter()

y_reg_pred_manual = X_test_lr @ linear_weights

manual_regression_prediction_time = (
    time.perf_counter() - start_time
)

print("Manual Linear Regression prediction time:",
      manual_regression_prediction_time,
      "seconds")
```

Implement the regression metrics ourselves

```python
def manual_mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))
```

```python
def manual_rmse(y_true, y_pred):
    return np.sqrt(np.mean((y_true - y_pred) ** 2))
```

```python
def manual_r2(y_true, y_pred):
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)

    return 1 - (ss_res / ss_tot)
```

```python
manual_mae_value = manual_mae(
    y_test_reg_manual,
    y_reg_pred_manual
)

manual_rmse_value = manual_rmse(
    y_test_reg_manual,
    y_reg_pred_manual
)

manual_r2_value = manual_r2(
    y_test_reg_manual,
    y_reg_pred_manual
)

print("Manual MAE :", manual_mae_value)
print("Manual RMSE:", manual_rmse_value)
print("Manual R2  :", manual_r2_value)
```

```python
manual_regression_results = {
    "Model": "From-Scratch Linear Regression",
    "MAE": manual_mae_value,
    "RMSE": manual_rmse_value,
    "R2": manual_r2_value,
    "Training Time": manual_regression_train_time,
    "Prediction Time": manual_regression_prediction_time
}

print(manual_regression_results)
```

```python
X_train_log = np.c_[
    np.ones(X_train_manual.shape[0]),
    X_train_manual
]

X_test_log = np.c_[
    np.ones(X_test_manual.shape[0]),
    X_test_manual
]

print("Classification training shape:", X_train_log.shape)
print("Classification testing shape:", X_test_log.shape)
```

```python
def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))
```

```python
def train_logistic_regression(X, y, learning_rate=0.01, epochs=2000):
    weights = np.zeros(X.shape[1])

    n_samples = X.shape[0]

    for epoch in range(epochs):

        # Linear combination
        z = X @ weights

        # Predicted probabilities
        probabilities = sigmoid(z)

        # Gradient
        gradient = (X.T @ (probabilities - y)) / n_samples

        # Update weights
        weights = weights - learning_rate * gradient

    return weights
```

```python
start_time = time.perf_counter()

logistic_weights = train_logistic_regression(
    X_train_log,
    y_train_cls_manual,
    learning_rate=0.01,
    epochs=2000
)

manual_classification_train_time = (
    time.perf_counter() - start_time
)

print(
    "Manual Logistic Regression training time:",
    manual_classification_train_time,
    "seconds"
)
```

```python
start_time = time.perf_counter()

y_cls_prob_manual = sigmoid(
    X_test_log @ logistic_weights
)

manual_classification_prediction_time = (
    time.perf_counter() - start_time
)

print(
    "Manual Logistic Regression prediction time:",
    manual_classification_prediction_time,
    "seconds"
)
```

```python
y_cls_pred_manual = (
    y_cls_prob_manual >= 0.5
).astype(int)

print(y_cls_pred_manual[:20])
```

Implement classification metrics manually

```python
def confusion_values(y_true, y_pred):
    tp = np.sum((y_true == 1) & (y_pred == 1))
    tn = np.sum((y_true == 0) & (y_pred == 0))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))

    return tp, tn, fp, fn
```

```python
def manual_accuracy(y_true, y_pred):
    tp, tn, fp, fn = confusion_values(y_true, y_pred)
    return (tp + tn) / (tp + tn + fp + fn)
```

```python
def manual_precision(y_true, y_pred):
    tp, tn, fp, fn = confusion_values(y_true, y_pred)

    if tp + fp == 0:
        return 0

    return tp / (tp + fp)
```

```python
def manual_recall(y_true, y_pred):
    tp, tn, fp, fn = confusion_values(y_true, y_pred)

    if tp + fn == 0:
        return 0

    return tp / (tp + fn)
```

```python
def manual_f1(y_true, y_pred):
    precision = manual_precision(y_true, y_pred)
    recall = manual_recall(y_true, y_pred)

    if precision + recall == 0:
        return 0

    return 2 * (precision * recall) / (precision + recall)
```

Calculate the results

```python
manual_accuracy_value = manual_accuracy(
    y_test_cls_manual,
    y_cls_pred_manual
)

manual_precision_value = manual_precision(
    y_test_cls_manual,
    y_cls_pred_manual
)

manual_recall_value = manual_recall(
    y_test_cls_manual,
    y_cls_pred_manual
)

manual_f1_value = manual_f1(
    y_test_cls_manual,
    y_cls_pred_manual
)

print("Accuracy :", manual_accuracy_value)
print("Precision:", manual_precision_value)
print("Recall   :", manual_recall_value)
print("F1-score :", manual_f1_value)
```

```python
manual_classification_results = {
    "Model": "From-Scratch Logistic Regression",
    "Accuracy": manual_accuracy_value,
    "Precision": manual_precision_value,
    "Recall": manual_recall_value,
    "F1": manual_f1_value,
    "Training Time": manual_classification_train_time,
    "Prediction Time": manual_classification_prediction_time
}

print(manual_classification_results)
```

Optimization

```python
learning_rates = [0.001, 0.01, 0.05, 0.1]
epochs_list = [1000, 2000, 5000]

optimization_results = []

for lr in learning_rates:
    for epochs in epochs_list:

        start_time = time.perf_counter()

        weights = train_logistic_regression(
            X_train_log,
            y_train_cls_manual,
            learning_rate=lr,
            epochs=epochs
        )

        training_time = time.perf_counter() - start_time

        probabilities = sigmoid(X_test_log @ weights)

        predictions = (probabilities >= 0.5).astype(int)

        accuracy_value = manual_accuracy(
            y_test_cls_manual,
            predictions
        )

        precision_value = manual_precision(
            y_test_cls_manual,
            predictions
        )

        recall_value = manual_recall(
            y_test_cls_manual,
            predictions
        )

        f1_value = manual_f1(
            y_test_cls_manual,
            predictions
        )

        optimization_results.append({
            "Learning Rate": lr,
            "Epochs": epochs,
            "Accuracy": accuracy_value,
            "Precision": precision_value,
            "Recall": recall_value,
            "F1": f1_value,
            "Training Time": training_time
        })

optimization_df = pd.DataFrame(optimization_results)

display(optimization_df)
```

```python
optimized_learning_rate = 0.05
optimized_epochs = 1000

start_time = time.perf_counter()

optimized_weights = train_logistic_regression(
    X_train_log,
    y_train_cls_manual,
    learning_rate=optimized_learning_rate,
    epochs=optimized_epochs
)

optimized_training_time = time.perf_counter() - start_time

# Prediction
start_time = time.perf_counter()

optimized_probabilities = sigmoid(
    X_test_log @ optimized_weights
)

optimized_predictions = (
    optimized_probabilities >= 0.5
).astype(int)

optimized_prediction_time = time.perf_counter() - start_time

print("Optimized training time:", optimized_training_time)
print("Optimized prediction time:", optimized_prediction_time)
```

```python
optimized_accuracy = manual_accuracy(
    y_test_cls_manual,
    optimized_predictions
)

optimized_precision = manual_precision(
    y_test_cls_manual,
    optimized_predictions
)

optimized_recall = manual_recall(
    y_test_cls_manual,
    optimized_predictions
)

optimized_f1 = manual_f1(
    y_test_cls_manual,
    optimized_predictions
)

print("Optimized Accuracy :", optimized_accuracy)
print("Optimized Precision:", optimized_precision)
print("Optimized Recall   :", optimized_recall)
print("Optimized F1       :", optimized_f1)
```

```python
optimized_classification_results = {
    "Model": "Optimized From-Scratch Logistic Regression",
    "Accuracy": optimized_accuracy,
    "Precision": optimized_precision,
    "Recall": optimized_recall,
    "F1": optimized_f1,
    "Training Time": optimized_training_time,
    "Prediction Time": optimized_prediction_time,
    "Learning Rate": optimized_learning_rate,
    "Epochs": optimized_epochs
}

print(optimized_classification_results)
```
